```python
import numpy as np

# Problema da Mochila (Blindagem de Ativos)
weights = np.array([12, 2, 1, 4, 1])   # Custo/Consumo de Memória dos ativos
values = np.array([4, 2, 1, 10, 2])    # Cobertura de Risco/Valor do ativo
max_weight = 15

pop_size = 10
num_genes = len(weights)
generations = 10
mutation_rate = 0.1

# Inicialização da População (Matriz Binária 10x5)
population = np.random.randint(0, 2, size=(pop_size, num_genes))

def calculate_fitness(ind):
    total_weight = np.sum(ind * weights)
    total_value = np.sum(ind * values)
    # Penalização: se estourar o peso maximo, fitness = 0
    if total_weight > max_weight:
        return 0
    return total_value

def tournament_selection(pop, fitnesses):
    # Seleciona 2 indices aleatorios e retorna o individuo de maior fitness
    idx1, idx2 = np.random.randint(0, len(pop), size=2)
    if fitnesses[idx1] > fitnesses[idx2]:
        return pop[idx1]
    else:
        return pop[idx2]

def crossover(parent1, parent2):
    point = np.random.randint(1, num_genes)
    child1 = np.concatenate([parent1[:point], parent2[point:]])
    child2 = np.concatenate([parent2[:point], parent1[point:]])
    return child1, child2

def mutate(ind):
    for i in range(num_genes):
        if np.random.rand() < mutation_rate:
            ind[i] = 1 - ind[i] # Inverte o bit (0 -> 1 ou 1 -> 0)
    return ind

# Loop Evolutivo
for g in range(generations):
    fitnesses = np.array([calculate_fitness(ind) for ind in population])

    melhor_idx = np.argmax(fitnesses)
    print(f"Geração {g:2d} | Melhor fitness: {fitnesses[melhor_idx]:2d} | Indivíduo: {population[melhor_idx]}")

    new_population = []
    for _ in range(pop_size // 2):
        p1 = tournament_selection(population, fitnesses)
        p2 = tournament_selection(population, fitnesses)
        c1, c2 = crossover(p1, p2)
        new_population.extend([mutate(c1), mutate(c2)])

    population = np.array(new_population)

# Resultado final
fitnesses = np.array([calculate_fitness(ind) for ind in population])
melhor_idx = np.argmax(fitnesses)
print("\n[LAB 02] MELHOR SOLUÇÃO FINAL")
print("Indivíduo:", population[melhor_idx])
print("Fitness (valor):", fitnesses[melhor_idx])
print("Peso usado:", np.sum(population[melhor_idx] * weights), "/", max_weight)
```
